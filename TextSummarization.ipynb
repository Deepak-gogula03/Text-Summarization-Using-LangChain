Text Summarization Using LangChain

```python
## Basic Prompt Summarization
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    AIMessage
)
```

```python
chat_messages=[
    SystemMessage(content="You are an expert Assistant with expertise in summarizing speeches"),
    HumanMessage(content="Please provide a short and concise summary of the following speech:\n TEXT: {speech}")
]
llm=ChatOpenAI(model_name="gpt-4o-mini")
```

```python
### total tokens
llm.get_num_tokens(speech)
```

```python
llm.invoke(chat_messages)
```

Prompt Template Text Summarization

```python
from langchain_classic.chains import LLMChain
from langchain_core.prompts import PromptTemplate
```

```python
generic_template = """
Write a Summary of the following speech.

Speech: {speech}

Translate the precise summary to {language}
"""

prompt = PromptTemplate(
    input_variables=["speech", "language"],
    template=generic_template
)
```

```python
complete_prompt=prompt.format(speech=speech,language='Hindi')
```

```python
complete_prompt
```

```python
llm.get_num_tokens(complete_prompt)
```

```python
llm_chain=LLMChain(llm=llm,prompt=prompt)
summary=llm_chain.run({'speech':speech,'language':'Hindi'})
```

```python
print(summary)
```

Summarizing Large Documents using Map Reduce

```python
from langchain_openai import ChatOpenAI
from langchain_classic.chains.summarize import load_summarize_chain
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
## provide the path of the pdf file/files
pdfreader=PdfReader("Martin_speech.pdf")
```

```python
from typing_extensions import Concatenate
## read text from the pdf
text=''
for i,page in enumerate(pdfreader.pages):
  content=page.extract_text()
  text+=content
```

```python
llm=ChatOpenAI(model_name="gpt-4o-mini")
```

```python
llm.get_num_tokens(text)
```

```python
### splitting the text
text_splitter=RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=20)
chunks=text_splitter.create_documents([text])
```

```python
chunks
```

```python
chain=load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    verbose=False
)
summary=chain.run(chunks)
```

```python
summary
```

Map Reduce With Custom Prompts

```python
chunks_prompt="""
Please summarize the below prompts
Speech: {text}
Summary :
"""
map_prompt_template=PromptTemplate(input_variables=['text'],template=chunks_prompt)
```

```python
final_combine_prompt="""
Provide a final summary of the entire speech with these important points.
Add a Generic Motivational Title,
Start the precise summary with an introduction and provide the summary in number points for the speech
Speech: {text}
"""
final_combine_prompt_template=PromptTemplate(input_variables=['text'],template=final_combine_prompt)
```

```python
summarize_chain=load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    map_prompt=map_prompt_template,
    combine_prompt=final_combine_prompt_template,
    verbose=False
)
output=summarize_chain.run(chunks)
```

```python
output
```

RefineChain for Summarization

```python
chain=load_summarize_chain(
    llm=llm,
    chain_type='refine',
    verbose=False
)
output_summary=chain.run(chunks)
```
